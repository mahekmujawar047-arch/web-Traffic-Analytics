```python
import pandas as pd
import os



file_name = "Website_Traffic_Data.csv"
df = pd.read_csv(file_name)



print("Dataset Preview:")
display(df.head())



print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
print("All Columns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Dataset Columns:")
print(df.columns.tolist())

print("\nFirst 10 Rows:")
display(df.head(10))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
display(df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


print("Columns:", df.columns.tolist())


print("\nMissing Values:")
print(df.isnull().sum())


duration_col = next((c for c in df.columns if "duration" in c.lower()), None)


print("\nSession Duration Summary:")
print(df[duration_col].describe())

df[duration_col].hist(bins=20)
plt.title("Distribution of Session Duration")
plt.xlabel("Session Duration")
plt.ylabel("Number of Sessions")
plt.show()


source_col = next((c for c in df.columns if "source" in c.lower()), None)


print("\nTraffic Source Counts:")
print(df[source_col].value_counts().head(10))


device_col = next((c for c in df.columns if "device" in c.lower()), None)


print("\nDevice Counts:")
print(df[device_col].value_counts())



print("\nTotal Sessions:", len(df))
print("Analysis completed successfully!")
```

```python
import matplotlib.pyplot as plt


df["Date_Key"] = pd.to_datetime(df["Date_Key"], errors="coerce")
daily_sessions = df.groupby("Date_Key")["Session_Id"].count()

plt.figure(figsize=(10, 4))
daily_sessions.plot()
plt.title("Website Sessions Over Time")
plt.xlabel("Date")
plt.ylabel("Number of Sessions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


source_counts = df["Source_Key"].value_counts().head(10)

plt.figure(figsize=(8, 4))
source_counts.plot(kind="bar")
plt.title("Top 10 Traffic Source Keys")
plt.xlabel("Source Key")
plt.ylabel("Number of Sessions")
plt.tight_layout()
plt.show()




device_counts = df["Device_Key"].value_counts()

plt.figure(figsize=(8, 4))
device_counts.plot(kind="bar")
plt.title("Sessions by Device Key")
plt.xlabel("Device Key")
plt.ylabel("Number of Sessions")
plt.tight_layout()
plt.show()

print("Graphs generated successfully!")
```
